```python
import requests
import pandas as pd
import time
from datetime import datetime
import matplotlib.pyplot as plt
import numpy as np

# Define keyword sets from your Table 1
keyword1_list = [
    "Wire Arc Additive Manufacturing",
    "Directed Energy Deposition",
    "DED",
    "Metal Additive Manufacturing",
    "Wire and Arc based Deposition"
]

keyword2_list = [
    "Life Cycle Assessment (LCA)",
    "Life Cycle Costing (LCC)",
    "Life Cycle Inventory (LCI)",
    "Life Cycle Impact Assessment (LCIA)",
    "Sustainability",
    "Green Manufacturing",
    "Environmental Impact",
    "Eco-efficiency",
    "Circular Economy",
    "Carbon Footprint",
    "Multi-Criteria Decision Analysis (MCDA)",
    "Optimization",
    "Uncertainty Quantification",
    "Material and Energy Flow Analysis (MEFA)",
    "Sensitivity Analysis",
    "SimaPro",
    "OpenLCA",
    "TOPSIS",
    "EPD",
    "Endpoint",
    "Midpoint",
    "Global Warming Potential (GWP)",
    "Energy Consumption"
]

def contains_keyword1(text, keyword1_list):
    """Check if text contains any of the WAAM/DED keywords"""
    if not text:
        return False
    text_lower = text.lower()
    for keyword in keyword1_list:
        if keyword.lower() in text_lower:
            return True
    return False

def contains_keyword2(text, keyword2):
    """Check if text contains the specific keyword2"""
    if not text:
        return False
    return keyword2.lower() in text.lower()

def check_exact_match(item, keyword1_list, keyword2):
    """Check if item contains any keyword1 AND the specific keyword2"""
    title = item.get('title', [''])[0]
    abstract = item.get('abstract', '')

    # Check if contains any WAAM/DED keyword AND the specific sustainability keyword
    return (contains_keyword1(title, keyword1_list) or contains_keyword1(abstract, keyword1_list)) and \
           (contains_keyword2(title, keyword2) or contains_keyword2(abstract, keyword2))

def clean_title(title):
    """Clean and format paper titles"""
    if isinstance(title, list):
        title = title[0]
    if not title or title == 'No title':
        return None
    return str(title).strip()

def clean_authors(authors_list):
    """Clean and format author names"""
    if not authors_list:
        return None

    authors = []
    for author in authors_list:
        given = author.get('given', '')
        family = author.get('family', '')
        author_name = f"{given} {family}".strip()
        if author_name:
            authors.append(author_name)

    return ", ".join(authors) if authors else None

def determine_article_type(item):
    """Determine if article is review or research"""
    article_type = item.get('type', '').lower()
    if 'review' in article_type:
        return "Review Article"
    elif any(t in article_type for t in ['research', 'article', 'journal-article']):
        return "Research Article"
    else:
        return "Other"

def query_crossref(start_year=2017):
    """Query the Crossref API for papers based on keyword combinations"""
    url = "https://api.crossref.org/works"
    all_papers = []
    seen_dois = set()

    for k2 in keyword2_list:
        # Search for each sustainability keyword combined with WAAM terms
        query = f"{k2}"
        print(f"Searching Crossref for sustainability keyword: {k2}")

        params = {
            "query": query,
            "filter": f"from-pub-date:{start_year},type:journal-article",
            "rows": 200,  # Increased to capture more results for filtering
            "select": "DOI,title,author,container-title,published,type,abstract"
        }

        try:
            response = requests.get(url, params=params, timeout=30)
            if response.status_code == 200:
                data = response.json()
                items = data['message'].get('items', [])

                # Filter items that contain any WAAM/DED keyword AND the current sustainability keyword
                filtered_items = [item for item in items if check_exact_match(item, keyword1_list, k2)]

                for item in filtered_items:
                    doi = item.get('DOI')
                    if not doi or doi in seen_dois:
                        continue

                    seen_dois.add(doi)

                    # Clean and extract data
                    title = clean_title(item.get('title'))
                    if not title:  # Skip if no valid title
                        continue

                    authors = clean_authors(item.get('author', []))
                    journal = item.get('container-title', [''])[0]

                    # Get publication year
                    published = item.get('published', {})
                    date_parts = published.get('date-parts', [[None]])[0]
                    year = date_parts[0] if date_parts and date_parts[0] else None

                    if not year or year < start_year:  # Skip if no valid year
                        continue

                    article_type = determine_article_type(item)

                    # Find which WAAM keyword was matched
                    matched_keyword1 = None
                    title_abstract = (title + " " + item.get('abstract', '')).lower()
                    for k1 in keyword1_list:
                        if k1.lower() in title_abstract:
                            matched_keyword1 = k1
                            break

                    paper_data = {
                        'Title': title,
                        'Year': year,
                        'Authors': authors if authors else "Not available",
                        'Journal': journal if journal else "Not available",
                        'DOI': doi,
                        'Article_Type': article_type,
                        'Database': 'Crossref',
                        'URL': f"https://doi.org/{doi}",
                        'Matched_WAAM_Keyword': matched_keyword1,
                        'Matched_Sustainability_Keyword': k2
                    }
                    all_papers.append(paper_data)

                print(f"  Found {len(filtered_items)} papers with WAAM/DED terms for '{k2}'")

            time.sleep(1)  # Be polite to the API

        except Exception as e:
            print(f"Error searching for {k2}: {e}")
            continue

    return all_papers

def query_pubmed_central(start_year=2017):
    """Query PubMed Central (Europe PMC) API"""
    url = "https://www.ebi.ac.uk/europepmc/webservices/rest/search"
    all_papers = []
    seen_pmids = set()

    for k2 in keyword2_list:
        # Search for each sustainability keyword
        query = f'("{k2}")'
        print(f"Searching PubMed Central for sustainability keyword: {k2}")

        params = {
            "query": query,
            "format": "json",
            "pageSize": 200,  # Increased to capture more results for filtering
            "sort": "YEAR desc"
        }

        try:
            response = requests.get(url, params=params, timeout=30)
            if response.status_code == 200:
                data = response.json()
                articles = data.get('resultList', {}).get('result', [])

                # Filter articles that contain any WAAM/DED keyword
                filtered_articles = []
                for article in articles:
                    title = article.get('title', '').strip()
                    abstract = article.get('abstractText', '')
                    if contains_keyword1(title, keyword1_list) or contains_keyword1(abstract, keyword1_list):
                        filtered_articles.append(article)

                for article in filtered_articles:
                    pmid = article.get('id')
                    if not pmid or pmid in seen_pmids:
                        continue

                    seen_pmids.add(pmid)

                    title = article.get('title', '').strip()
                    if not title or title == 'No title':
                        continue

                    # Get authors
                    authors_list = []
                    for author in article.get('authorList', {}).get('author', []):
                        name = author.get('fullName', '').strip()
                        if name:
                            authors_list.append(name)
                    authors = ", ".join(authors_list) if authors_list else "Not available"

                    year = article.get('pubYear')
                    if not year or int(year) < start_year:
                        continue

                    # Determine article type
                    pub_type = article.get('pubType', '').lower()
                    is_review = 'review' in pub_type
                    article_type = "Review Article" if is_review else "Research Article"

                    # Find which WAAM keyword was matched
                    matched_keyword1 = None
                    title_abstract = (title + " " + article.get('abstractText', '')).lower()
                    for k1 in keyword1_list:
                        if k1.lower() in title_abstract:
                            matched_keyword1 = k1
                            break

                    paper_data = {
                        'Title': title,
                        'Year': int(year),
                        'Authors': authors,
                        'Journal': article.get('journalTitle', 'Not available'),
                        'DOI': article.get('doi', 'Not available'),
                        'Article_Type': article_type,
                        'Database': 'PubMed Central',
                        'URL': article.get('pdfUrl', f"https://europepmc.org/article/MED/{pmid}"),
                        'Matched_WAAM_Keyword': matched_keyword1,
                        'Matched_Sustainability_Keyword': k2
                    }
                    all_papers.append(paper_data)

                print(f"  Found {len(filtered_articles)} papers with WAAM/DED terms for '{k2}'")

            time.sleep(0.5)

        except Exception as e:
            print(f"Error searching for {k2}: {e}")
            continue

    return all_papers

def clean_dataframe(df):
    """Clean the dataframe by removing nonsense rows and duplicates"""
    if df.empty:
        return df

    # Remove rows with missing critical information
    df_clean = df.dropna(subset=['Title', 'Year'])

    # Remove duplicates based on DOI and Title
    df_clean = df_clean.drop_duplicates(subset=['DOI'], keep='first')
    df_clean = df_clean.drop_duplicates(subset=['Title'], keep='first')

    # Filter out nonsense titles
    nonsense_keywords = ['test', 'example', 'conference', 'abstract', 'untitled']
    mask = ~df_clean['Title'].str.lower().str.contains('|'.join(nonsense_keywords), na=False)
    df_clean = df_clean[mask]

    # Ensure Year is integer and within range
    current_year = datetime.now().year
    df_clean = df_clean[(df_clean['Year'] >= 2017) & (df_clean['Year'] <= current_year)]

    # Sort by year (descending) and title
    df_clean = df_clean.sort_values(['Year', 'Title'], ascending=[False, True])

    # Reset index
    df_clean = df_clean.reset_index(drop=True)

    return df_clean

def create_bar_chart(df):
    """Create a bar chart showing publications per year"""
    if df.empty:
        print("No data available for chart")
        return

    # Count publications per year
    year_counts = df['Year'].value_counts().sort_index()

    # Create figure and axis
    plt.figure(figsize=(12, 6))

    # Create bar chart
    bars = plt.bar(year_counts.index.astype(str), year_counts.values,
                   color='skyblue', edgecolor='black', alpha=0.7)

    # Add value labels on bars
    for bar in bars:
        height = bar.get_height()
        plt.text(bar.get_x() + bar.get_width()/2., height,
                f'{int(height)}', ha='center', va='bottom')

    # Customize chart
    plt.title('Number of WAAM/DED LCA/LCC Publications per Year (2017-Present)',
              fontsize=14, fontweight='bold')
    plt.xlabel('Publication Year', fontsize=12)
    plt.ylabel('Number of Publications', fontsize=12)
    plt.xticks(rotation=45)
    plt.grid(axis='y', alpha=0.3)

    # Adjust layout
    plt.tight_layout()

    # Save the chart
    plt.savefig('publications_per_year.png', dpi=300, bbox_inches='tight')
    plt.show()

    print(f"Bar chart saved as 'publications_per_year.png'")

def create_keyword_analysis(df):
    """Create analysis of which keywords were most effective"""
    if df.empty:
        return

    print(f"\n🔑 KEYWORD ANALYSIS:")
    print("=" * 50)

    # WAAM keyword effectiveness
    waam_counts = df['Matched_WAAM_Keyword'].value_counts()
    print(f"WAAM/DED Keyword Effectiveness:")
    for keyword, count in waam_counts.items():
        print(f"  - {keyword}: {count} papers")

    # Sustainability keyword effectiveness
    sustainability_counts = df['Matched_Sustainability_Keyword'].value_counts()
    print(f"\nSustainability Keyword Effectiveness:")
    for keyword, count in sustainability_counts.head(10).items():  # Top 10
        print(f"  - {keyword}: {count} papers")

# Main execution
print("Starting comprehensive literature search...")
print(f"WAAM/DED keywords: {len(keyword1_list)}")
print(f"Sustainability/LCA keywords: {len(keyword2_list)}")
print(f"Time period: 2017 to present")
print("🔍 IMPORTANT: Results MUST contain at least one WAAM/DED keyword AND one sustainability keyword")
print("=" * 80)

# Query both databases
print("\n🔍 Searching academic databases...")
all_results = []
all_results.extend(query_crossref(2017))
all_results.extend(query_pubmed_central(2017))

# Create DataFrame and clean data
df = pd.DataFrame(all_results)
print(f"\n📊 Found {len(df)} records before cleaning")

# Clean the data
df_clean = clean_dataframe(df)
print(f"📊 After cleaning: {len(df_clean)} valid publications")

# Display and save results
if not df_clean.empty:
    # Save to CSV with proper columns
    csv_filename = "WAAM_LCA_Literature_Review.csv"
    df_clean.to_csv(csv_filename, index=False, encoding='utf-8')
    print(f"💾 Results saved to {csv_filename}")

    # Display the cleaned data
    print(f"\n📋 First 10 publications:")
    print("=" * 100)
    display(df_clean.head(10))

    # Summary statistics
    print(f"\n📈 SUMMARY STATISTICS:")
    print("=" * 50)
    print(f"Total unique publications: {len(df_clean)}")
    print(f"\nPublications by type:")
    print(df_clean['Article_Type'].value_counts())
    print(f"\nPublications by database:")
    print(df_clean['Database'].value_counts())

    # Keyword analysis
    create_keyword_analysis(df_clean)

    # Create bar chart
    print(f"\n📊 Creating visualization...")
    create_bar_chart(df_clean)

    # Additional analysis
    print(f"\n📅 Publication years: {sorted(df_clean['Year'].unique())}")

    # Save detailed summary
    summary_filename = "search_summary.txt"
    with open(summary_filename, 'w') as f:
        f.write("WAAM/DED LCA/LCC Literature Search Summary\n")
        f.write("=" * 50 + "\n")
        f.write(f"Search date: {datetime.now().strftime('%Y-%m-%d %H:%M')}\n")
        f.write(f"Total publications found: {len(df_clean)}\n")
        f.write(f"Time period: 2017-{datetime.now().year}\n")
        f.write(f"Search requirement: Must contain at least one WAAM/DED keyword AND one sustainability keyword\n")
        f.write(f"WAAM/DED keywords used: {keyword1_list}\n")
        f.write(f"Publication types:\n")
        for article_type, count in df_clean['Article_Type'].value_counts().items():
            f.write(f"  - {article_type}: {count}\n")
        f.write(f"Databases searched: Crossref, PubMed Central\n")
        f.write(f"Years covered: {sorted(df_clean['Year'].unique())}\n")
        f.write(f"\nMost effective WAAM keywords:\n")
        for keyword, count in df_clean['Matched_WAAM_Keyword'].value_counts().items():
            f.write(f"  - {keyword}: {count}\n")

    print(f"📝 Search summary saved to {summary_filename}")

else:
    print("❌ No valid publications found after cleaning.")
    print("💡 Try adjusting your search terms or checking the API connectivity.")

print("\n✅ Search completed!")
```

```python
import requests
import pandas as pd
import time
from datetime import datetime
import matplotlib.pyplot as plt

# =================================================================
# TIME WINDOW  —  2017 to PRESENT (dynamic)
# =================================================================
START_DATE   = "2017-01-01"
TODAY        = datetime.now()
END_DATE     = TODAY.strftime("%Y-%m-%d")      # today
CURRENT_YEAR = TODAY.year                      # e.g. 2026

print(f"Search window: {START_DATE} to {END_DATE}")

# =================================================================
# KEYWORD SETS  (Table 1 – unchanged fundamentals)
# =================================================================
keyword1_list = [
    "Wire Arc Additive Manufacturing",
    "Directed Energy Deposition",
    "DED",
    "Metal Additive Manufacturing",
    "Wire and Arc based Deposition",
]

keyword2_list = [
    "Life Cycle Assessment (LCA)",
    "Life Cycle Costing (LCC)",
    "Life Cycle Inventory (LCI)",
    "Life Cycle Impact Assessment (LCIA)",
    "Sustainability",
    "Green Manufacturing",
    "Environmental Impact",
    "Eco-efficiency",
    "Circular Economy",
    "Carbon Footprint",
    "Multi-Criteria Decision Analysis (MCDA)",
    "Optimization",
    "Uncertainty Quantification",
    "Material and Energy Flow Analysis (MEFA)",
    "Sensitivity Analysis",
    "SimaPro",
    "OpenLCA",
    "TOPSIS",
    "EPD",
    "Endpoint",
    "Midpoint",
    "Global Warming Potential (GWP)",
    "Energy Consumption",
]

# =================================================================
# MATCHING HELPERS  (original loose substring matching – unchanged)
# =================================================================
def contains_keyword1(text, keyword1_list):
    if not text:
        return False
    text_lower = text.lower()
    for keyword in keyword1_list:
        if keyword.lower() in text_lower:
            return True
    return False

def contains_keyword2(text, keyword2):
    if not text:
        return False
    return keyword2.lower() in text.lower()

def check_exact_match(item, keyword1_list, keyword2):
    title = item.get('title', [''])[0]
    abstract = item.get('abstract', '')
    return (contains_keyword1(title, keyword1_list) or contains_keyword1(abstract, keyword1_list)) and \
           (contains_keyword2(title, keyword2) or contains_keyword2(abstract, keyword2))

def clean_title(title):
    if isinstance(title, list):
        title = title[0]
    if not title or title == 'No title':
        return None
    return str(title).strip()

def clean_authors(authors_list):
    if not authors_list:
        return None
    authors = []
    for author in authors_list:
        given = author.get('given', '')
        family = author.get('family', '')
        name = f"{given} {family}".strip()
        if name:
            authors.append(name)
    return ", ".join(authors) if authors else None

def determine_article_type(item):
    article_type = item.get('type', '').lower()
    if 'review' in article_type:
        return "Review Article"
    elif any(t in article_type for t in ['research', 'article', 'journal-article']):
        return "Research Article"
    return "Other"

# =================================================================
# STAGE 1 – CROSSREF (automated)
# =================================================================
def query_crossref(start_date=START_DATE, end_date=END_DATE):
    url = "https://api.crossref.org/works"
    all_papers = []
    seen_dois = set()

    for k2 in keyword2_list:
        print(f"[Crossref] Searching: {k2}")
        params = {
            "query": f"{k2}",
            "filter": f"from-pub-date:{start_date},"
                      f"until-pub-date:{end_date},"
                      f"type:journal-article",
            "rows": 200,
            "select": "DOI,title,author,container-title,published,type,abstract"
        }
        try:
            response = requests.get(url, params=params, timeout=30)
            if response.status_code == 200:
                items = response.json()['message'].get('items', [])
                filtered = [it for it in items
                            if check_exact_match(it, keyword1_list, k2)]

                for item in filtered:
                    doi = item.get('DOI')
                    if not doi or doi in seen_dois:
                        continue
                    seen_dois.add(doi)

                    title = clean_title(item.get('title'))
                    if not title:
                        continue

                    published = item.get('published', {})
                    parts = published.get('date-parts', [[None]])[0]
                    year = parts[0] if parts and parts[0] else None
                    if not year or year < 2017 or year > CURRENT_YEAR:
                        continue

                    matched_k1 = None
                    ta = (title + " " + item.get('abstract', '')).lower()
                    for k1 in keyword1_list:
                        if k1.lower() in ta:
                            matched_k1 = k1
                            break

                    all_papers.append({
                        'Title': title,
                        'Year': year,
                        'Authors': clean_authors(item.get('author', [])) or "Not available",
                        'Journal': item.get('container-title', [''])[0] or "Not available",
                        'DOI': doi,
                        'Article_Type': determine_article_type(item),
                        'Database': 'Crossref',
                        'URL': f"https://doi.org/{doi}",
                        'Matched_WAAM_Keyword': matched_k1,
                        'Matched_Sustainability_Keyword': k2,
                    })

                print(f"          -> {len(filtered)} candidate records")
            time.sleep(1)
        except Exception as e:
            print(f"          [warn] {k2}: {e}")
            continue
    return all_papers

# =================================================================
# STAGE 2 – EUROPE PMC (automated)
# =================================================================
def query_pubmed_central(start_year=2017, end_year=CURRENT_YEAR):
    url = "https://www.ebi.ac.uk/europepmc/webservices/rest/search"
    all_papers = []
    seen_pmids = set()

    for k2 in keyword2_list:
        print(f"[Europe PMC] Searching: {k2}")
        params = {
            "query": f'("{k2}")',
            "format": "json",
            "pageSize": 200,
            "sort": "YEAR desc"
        }
        try:
            r = requests.get(url, params=params, timeout=30)
            if r.status_code == 200:
                articles = r.json().get('resultList', {}).get('result', [])
                filtered = []
                for a in articles:
                    title = a.get('title', '').strip()
                    abstract = a.get('abstractText', '')
                    if contains_keyword1(title, keyword1_list) or \
                       contains_keyword1(abstract, keyword1_list):
                        filtered.append(a)

                for a in filtered:
                    pmid = a.get('id')
                    if not pmid or pmid in seen_pmids:
                        continue
                    seen_pmids.add(pmid)

                    title = a.get('title', '').strip()
                    if not title:
                        continue

                    year = a.get('pubYear')
                    try:
                        year = int(year)
                    except (TypeError, ValueError):
                        continue
                    if not (start_year <= year <= end_year):
                        continue

                    matched_k1 = None
                    ta = (title + " " + a.get('abstractText', '')).lower()
                    for k1 in keyword1_list:
                        if k1.lower() in ta:
                            matched_k1 = k1
                            break

                    pub_type = a.get('pubType', '').lower()
                    atype = "Review Article" if 'review' in pub_type else "Research Article"

                    all_papers.append({
                        'Title': title,
                        'Year': year,
                        'Authors': a.get('authorString', 'Not available'),
                        'Journal': a.get('journalTitle', 'Not available'),
                        'DOI': a.get('doi', 'Not available'),
                        'Article_Type': atype,
                        'Database': 'Europe PMC',
                        'URL': a.get('pdfUrl', f"https://europepmc.org/article/MED/{pmid}"),
                        'Matched_WAAM_Keyword': matched_k1,
                        'Matched_Sustainability_Keyword': k2,
                    })
                print(f"          -> {len(filtered)} candidate records")
            time.sleep(0.5)
        except Exception as e:
            print(f"          [warn] {k2}: {e}")
            continue
    return all_papers

# =================================================================
# STAGE 3 – MANUAL SCOPUS + WEB OF SCIENCE
# =================================================================
# Same keyword combinations from Table 1 were run manually in Scopus
# and Web of Science to capture conference proceedings and reviews
# not indexed as journal articles by Crossref.
#
# Export Scopus/WoS results as CSV, rename columns to:
#   Title, Year, Authors, Journal, DOI, Article_Type
# Then uncomment and adjust the loader below.
#
# def load_manual(path, source_label):
#     df = pd.read_csv(path)
#     df = df.rename(columns={
#         "Article Title": "Title",
#         "Publication Year": "Year",
#         "Author Full Names": "Authors",
#         "Source Title": "Journal",
#         "DOI": "DOI",
#         "Document Type": "Article_Type",
#     })
#     df["Database"] = source_label
#     df["URL"] = df["DOI"].apply(
#         lambda d: f"https://doi.org/{d}" if isinstance(d, str) and d else ""
#     )
#     df["Matched_WAAM_Keyword"] = "manual"
#     df["Matched_Sustainability_Keyword"] = "manual"
#     return df[["Title","Year","Authors","Journal","DOI",
#                "Article_Type","Database","URL",
#                "Matched_WAAM_Keyword","Matched_Sustainability_Keyword"]]
#
# scopus_papers = load_manual("scopus_export.csv", "Scopus").to_dict("records")
# wos_papers    = load_manual("wos_export.csv",    "Web of Science").to_dict("records")

scopus_papers = []
wos_papers    = []

# =================================================================
# STAGE 4 – CLEAN + DEDUP
# =================================================================
def clean_dataframe(df):
    if df.empty:
        return df
    df_clean = df.dropna(subset=['Title', 'Year'])

    # Deduplication: DOI + title only (no author matching)
    df_clean = df_clean.drop_duplicates(subset=['DOI'], keep='first')
    df_clean = df_clean.drop_duplicates(subset=['Title'], keep='first')

    # No "conference" title filter — conference proceedings retained
    df_clean = df_clean[(df_clean['Year'] >= 2017) &
                        (df_clean['Year'] <= CURRENT_YEAR)]
    df_clean = df_clean.sort_values(['Year', 'Title'], ascending=[False, True])
    return df_clean.reset_index(drop=True)

# =================================================================
# STAGE 5 – FIGURE 3
# =================================================================
def create_bar_chart(df):
    if df.empty:
        print("No data for chart")
        return
    year_counts = df['Year'].value_counts().sort_index()
    plt.figure(figsize=(12, 6))
    bars = plt.bar(year_counts.index.astype(str), year_counts.values,
                   color='skyblue', edgecolor='black', alpha=0.8)
    for bar in bars:
        h = bar.get_height()
        plt.text(bar.get_x() + bar.get_width()/2., h, f'{int(h)}',
                 ha='center', va='bottom')
    plt.title(f'Annual Number of WAAM/DED LCA/LCC Publications '
              f'(2017 – {CURRENT_YEAR})', fontsize=13, fontweight='bold')
    plt.xlabel('Year'); plt.ylabel('Number of Publications')
    plt.xticks(rotation=45); plt.grid(axis='y', alpha=0.3)
    plt.tight_layout()
    plt.savefig('publications_per_year.png', dpi=300, bbox_inches='tight')
    plt.show()
    print("Saved publications_per_year.png")

# =================================================================
# MAIN
# =================================================================
print("=" * 70)
print("WAAM/DED LCA/LCC Systematic Literature Search")
print("Automated: Crossref + Europe PMC")
print("Manual:    Scopus + Web of Science")
print(f"Window:    {START_DATE} to {END_DATE}")
print("=" * 70)

print("\n[Stage 1] Crossref ...")
crossref_papers = query_crossref()
print(f"          Crossref total: {len(crossref_papers)}")

print("\n[Stage 2] Europe PMC ...")
epmc_papers = query_pubmed_central()
print(f"          Europe PMC total: {len(epmc_papers)}")

print("\n[Stage 3] Manual Scopus + Web of Science ...")
print(f"          Scopus total:         {len(scopus_papers)}")
print(f"          Web of Science total: {len(wos_papers)}")

all_results = crossref_papers + epmc_papers + scopus_papers + wos_papers
print(f"\n[Stage 4] Total before cleaning: {len(all_results)}")

df = pd.DataFrame(all_results)
df_clean = clean_dataframe(df)
print(f"          Total after dedup + screening: {len(df_clean)}")

# ---- PRISMA-style reporting ----
print("\n" + "=" * 70)
print("PRISMA-STYLE FLOW")
print("=" * 70)
print(f"Crossref (automated):           {len(crossref_papers)}")
print(f"Europe PMC (automated):         {len(epmc_papers)}")
print(f"Scopus (manual supplement):     {len(scopus_papers)}")
print(f"Web of Science (manual):        {len(wos_papers)}")
print(f"After DOI/title dedup + screen: {len(df_clean)}")

if not df_clean.empty:
    df_clean.to_csv("WAAM_LCA_Literature_Review.csv",
                    index=False, encoding='utf-8')

    print("\nBy database:")
    print(df_clean['Database'].value_counts().to_string())
    print("\nBy article type:")
    print(df_clean['Article_Type'].value_counts().to_string())

    print("\nYearly distribution (Figure 3):")
    yearly = df_clean['Year'].value_counts().sort_index()
    print(yearly.to_string())

    create_bar_chart(df_clean)

    with open("search_summary.txt", "w") as f:
        f.write("WAAM/DED LCA/LCC Literature Search Summary\n")
        f.write("=" * 50 + "\n")
        f.write(f"Search date: {datetime.now():%Y-%m-%d %H:%M}\n")
        f.write("Automated: Crossref + Europe PMC\n")
        f.write("Manual:    Scopus + Web of Science\n")
        f.write(f"Window:    {START_DATE} to {END_DATE}\n\n")
        f.write(f"Crossref:       {len(crossref_papers)}\n")
        f.write(f"Europe PMC:     {len(epmc_papers)}\n")
        f.write(f"Scopus:         {len(scopus_papers)}\n")
        f.write(f"Web of Science: {len(wos_papers)}\n")
        f.write(f"Final:          {len(df_clean)}\n\n")
        f.write("Yearly distribution:\n")
        for y, n in yearly.items():
            f.write(f"  {y}: {n}\n")

print("\nDone.")
```
